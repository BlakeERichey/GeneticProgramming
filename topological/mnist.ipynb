```python
import numpy as np
import torch
from torchvision import datasets, transforms
from tpcomp.inference import Topology
from tpcomp.solver import Genome

# ----------------------------
# 1. Load MNIST Dataset
# ----------------------------
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Lambda(lambda x: x.view(-1))  # Flatten to 1D (784)
])

# Load training and test sets
train_dataset = datasets.MNIST(root='E\\data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='E\\data', train=False, download=True, transform=transform)

# Use a subset (e.g., 1000 test images) for testing
n_test = 1000
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=1, shuffle=False)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=1, shuffle=False)
```

```python
n_train = 100
```

```python
for attempt in range(100):
    try:
        genome = Genome(n_obs=255, n_act=11) # 256 = uint8 for Grayscale
        top = Topology(genome) #return a randomly initialized topology
        
        correct = 0
        total = 0
        
        for i, (data, target) in enumerate(train_loader):
            if i >= n_train:
                break
        
            # Flatten image (already 1024 in size)
            image = data.squeeze().numpy()  # Shape: (1024,)
            image = (image * 255).astype(np.uint8)  # Convert to uint8 (0-255)
        
            # Run through topology
            top.stage(image)
            response = top.response()
            top.reset()
        
            # Get predicted class
            # response is a Counter-like object; most_common() returns list of (class_id, count)
            predicted_class = response.most_common()[0][0]
            # Compare with true label
            true_class = target.item()
            print(predicted_class, true_class)
        
            if predicted_class == true_class:
                correct += 1
            else:
                break
            total += 1
        
            if i % 1 == 0:
                print(f"Tested {i}/{n_test}, Accuracy so far: {correct / total:.3f}")    
    except AssertionError:
        pass
```
